### Exercise 1(a) (10 points)

Create a class called `DataFrameProcessor` that will handle basic operations on a pandas DataFrame. Your class should:

1. Initialize with a pandas DataFrame.
2. Have a method get_column_stats(column_name) that returns the mean, median, and standard deviation of the specified column.
3. Have a method filter_rows(column_name, threshold) that returns a new DataFrame with rows where the values in the specified column are greater than the given threshold.
4. Have a method add_new_column(column_name, data) that adds a new column with the specified name and data to the DataFrame.

````python
class DataFrameProcessor:
    def __init__(self, dataframe):
        """Initialize the DataFrameProcessor with a pandas DataFrame."""
        self.dataframe = dataframe

    def get_column_stats(self, column_name):
        """Return the mean, median, and standard deviation of the specified column."""
        mean_value = self.dataframe[column_name].mean()
        median_value = self.dataframe[column_name].median()
        std_value = self.dataframe[column_name].std()

        return {
            "mean": mean_value,
            "median": median_value,
            "std": std_value
        }

    def filter_rows(self, column_name, threshold):
        """Return rows where the specified column is greater than the threshold."""
        return self.dataframe[self.dataframe[column_name] > threshold]

    def add_new_column(self, column_name, data):
        """Add a new column to the DataFrame."""
        self.dataframe[column_name] = data
````

Data pre-processing with OOP. Create a Python class called `DataPreprocessor` that:

1. Takes a pandas DataFrame as input.
2. Provides methods to:
   - Handle missing values by filling them with the column mean.
   - Normalize numerical columns between 0 and 1.
3. Returns the cleaned DataFrame when called with `.transform()`

Consider `df` as DataFrame to test the class.

```
data = {
    "age": [25, np.nan, 35, 40, np.nan],
    "income": [50000, 60000, np.nan, 80000, 75000],
    "gender": ["M", "F", "M", "F", "M"]
}
df = pd.DataFrame(data)
```

````python
class DataPreprocessor:
    def __init__(self, dataframe):
        """Initializes the DataPreprocessor with the given DataFrame."""
        self.dataframe = dataframe

    def filling_missing_values(self):
        """Fills missing values in numeric columns with the column mean."""
        for col in self.dataframe.select_dtypes(include=[np.number]).columns:
            self.dataframe[col] = self.dataframe[col].fillna(
                self.dataframe[col].mean()
            )

    def normalize(self):
        """Normalizes numeric columns in the DataFrame to the range [0, 1]."""
        for col in self.dataframe.select_dtypes(include=[np.number]).columns:
            col_min = self.dataframe[col].min()
            col_max = self.dataframe[col].max()

            self.dataframe[col] = (
                (self.dataframe[col] - col_min) / (col_max - col_min)
            )

    def transform(self):
        """Applies preprocessing and returns the cleaned DataFrame."""
        self.filling_missing_values()
        self.normalize()
        return self.dataframe                
````
